### 🟠 Nível 3: Curiosidades e Estatísticas Simples

Vamos usar funções matemáticas básicas para entender a lista.

**Exercício 7:**
Qual é a **média das notas** (`Rating`) de todos os 250 filmes? E qual é a média dos anos de lançamento?

*Dica: Use o comando `.mean()` na coluna desejada.*

```python
# Seu código aqui
# Média das notas
media_notas = df["Rating"].mean()

# Média dos anos de lançamento
media_anos = df["Year"].mean()

# Mostrar os resultados
print("Média das notas:", media_notas)
print("Média dos anos:", media_anos)
```

**Exercício 8:**
Quem é o filme com a **maior duração** da lista? E qual é essa duração?
*(Dica: Como a duração está em texto, ex: '2h 22m', você pode ordenar a coluna 'Duration' e ver qual fica no topo ou no final).*

```python
# Seu código aqui
# Ordenar pela duração, da maior para a menor
df_ordenado = df.sort_values(by="Duration", ascending=False)

# Mostrar o filme com maior duração
df_ordenado[["Title", "Duration"]].head(1)
```

### 🔴 Nível 4: Um pouco de Visualização (Gráficos)

Uma imagem vale mais que mil palavras. Vamos criar gráficos simples.

**Exercício 9:**
Crie um **gráfico de barras** mostrando a contagem de filmes por ano (`Year`).
*Pergunta:* Teve algum ano que teve muitos filmes na lista?

*Dica: Use `df['Year'].value_counts().head(10).plot(kind='bar')`*

```python
# Seu código aqui
# Contar quantos filmes existem por ano
contagem_anos = df["Year"].value_counts().head(10)

# Criar gráfico de barras
contagem_anos.plot(kind="bar")
```

**Exercício 10:**
Crie um **histograma** para ver como as notas estão distribuídas. A maioria dos filmes tem nota 8, 8.5 ou 9?

*Dica: Use `df['Rating'].plot(kind='hist', bins=20)`*

```python
# Seu código aqui
import matplotlib.pyplot as plt

# Criar histograma das notas
df["Rating"].plot(kind="hist", bins=20)

# Adicionar título e rótulos
plt.title("Distribuição das notas dos filmes")
plt.xlabel("Nota (Rating)")
plt.ylabel("Quantidade de filmes")

plt.show() 

df["Rating"].value_counts().sort_index()
```
